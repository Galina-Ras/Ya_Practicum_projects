**Описание данных**

    Файл moscow_places.csv:

* **_name_** — название заведения;

* **_address_** — адрес заведения;

* **_category_** — категория заведения, например «кафе», «пиццерия» или «кофейня»;

* **_hours_** — информация о днях и часах работы;

* **_lat_** — широта географической точки, в которой находится заведение;

* **_lng_** — долгота географической точки, в которой находится заведение;

* **_rating_** — рейтинг заведения по оценкам пользователей в Яндекс Картах (высшая оценка — 5.0);

* **_price_** — категория цен в заведении, например «средние», «ниже среднего», «выше среднего» и так далее;

* **_avg_bill_** — строка, которая хранит среднюю стоимость заказа в виде диапазона, например:

    o	«Средний счёт: 1000–1500 ₽»;

    o	«Цена чашки капучино: 130–220 ₽»;

    o	«Цена бокала пива: 400–600 ₽».

    o	и так далее;


* **_middle_avg_bill_** — число с оценкой среднего чека, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Средний счёт»:

    o	Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
    
    o	Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
    
    o	Если значения нет или оно не начинается с подстроки «Средний счёт», то в столбец ничего не войдёт.
    

* **_middle_coffee_cup_** — число с оценкой одной чашки капучино, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Цена одной чашки капучино»:

    o Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
  
    o	Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.

    o	Если значения нет или оно не начинается с подстроки «Цена одной чашки капучино», то в столбец ничего не войдёт.


* **_chain_** — число, выраженное 0 или 1, которое показывает, является ли заведение сетевым (для маленьких сетей могут встречаться ошибки); 

* **_district_** — административный район, в котором находится заведение, например Центральный административный округ;

* **_seats_** — количество посадочных мест.

### Построим фоновую картограмму (хороплет) со средним рейтингом заведений каждого района<a id="three_eight"></a>

```python
ratings_in_districts_map = df.groupby('district', as_index=False)['rating'].agg('mean').sort_values('rating', ascending=False)
display(ratings_in_districts_map)
```

Необходимые карта и хороплет уже импортированы в начале, теперь загружаем JSON-файл с границами округов Москвы:

```python
state_geo = 'https://code.s3.yandex.net/data-analyst/admin_level_geomap.geojson'
moscow_lat, moscow_lng = 55.751244, 37.618423  # центр Москвы
print(state_geo)
```

### Отобразим все заведения датасета на карте с помощью кластеров средствами библиотеки folium<a id="three_nine"></a>

Создаём пустой кластер, добавляем его на карту:

```python
marker_cluster = MarkerCluster().add_to(moscow_map)

# Напишем функцию, которая принимает строку датафрейма,создаёт маркер в текущей точке и добавляет его в кластер marker_cluster

def create_marker(row):
    Marker([row['lat'], row['lng']],
           popup=f"{row['name']} {row['rating']}",
          ).add_to(marker_cluster)

# применяем функцию create_clusters() к каждой строке датафрейма:
df.apply(create_marker, axis=1)

# выводим карту:
moscow_map
```

- В ЦАО в целом средний рейтинг заведений выше всех, 4.4. А вот в ЮВАО в среднем средний рейтинг заведений ниже - 4.1, хотя рейтинг булочных там почти выше среднего рейтинга всех булочных в остальных округах Москвы;

- У пиццерий, кофеен и булочных примерно одинаковый рейтинг;

- Рестораны, как и можно догадаться, имеют наивысший рейтинг именно в ЦАО.

    8) Отобразив все заведения датасета на карте, мы заметили, что основная часть заведений общепита располагается в самом центре города.

    9) Также мы определили топ-15 улиц по количеству заведений:

- Среди топ-15 улиц Москвы больше всего заведений общепита расположено на проспекте Мира, где преобладают такие категории заведений (183 заведения), как кафе и рестораны;
    
- Далее много заведений располагается на улице Профсоюзная (117), однако это в 1.5 раза меньше заведений, чем у улицы-лидера; 
    
- Меньше всего заведений располагается на Волгоградском проспекте  - всего 41 заведение;

- На 9 улицах из топ-15 лидерами среди популярных заведений являются кафе. 
     
    10) Было обнаружено, что по Москве есть аж 463 улицы, на которых находится только один объект общепита:
 
- В основном, в единственном экземпляре на улице находятся кафе (их 162 штуки);
    
- Затем идут рестораны и кофейни;
    
- Но меньше всего булочных - их 7 штук из всего датасета;
    
- Бары же, пабы, пиццерии, заведения быстрого питания открываются там чаще всего, где есть большая проходимость клиентов. Иначе они просто не окупятся.

    11) Определено, что самый высокий показатель среднего чека по заведеням Москвы -  в ЦАО и ЗАО. Самый же маленький средний чек в ЮВАО,и это значение в 2 раза меньше, чем у двух предыдущих округах. 
 
    
    
3. <b> Проведен анализ кафеен в Москве: <b>      

Проанализировав кофейни в Москве, мы обнаружили следующее:
    
    1) Всего в Москве примерно 1331 кофейня: 
- Больше всего кофеен расположено в ЦАО (407 штук) - это 1/3 от всех кофеен по всей Москве;

- Далее, большое количество кофеен локализовано в САО, СВАО и ЗАО (180, 155 и 140 соответственно);

- Меньше всего кофеен имеется в СЗАО - всего 60 штук. Это почти 7 раз меньше, чем в ЦАО.

    2) 56 из всех кофеен Москвы являются круглосуточными:
    
- Снова больше всего кофеен расположено в ЦАО (25 штук) - это почти половина от всех круглосуточных кофеен  по всей Москве. Это и понятно, ведь людям по нраву засидеться до позднего вечера в центре столицы;
    
- Во всех остальных районах Москвы кофеен значительно меньше, но меньше всего в СВАО, СЗАО и ЮВАО - там по 2 и 1 кофеен; 
    
- Также на построенной карте видно, что в ЮАО нет круглосточных кафеен вовсе (согласно имеющемуся датасету).

    3) Рейтинги кофеен:

- Самые высокие показатели среднего рейтинга кофеен в ЦАО и СЗАО (от 4.3);
    
- Средний рейтинг заведений до 4.2 присущ заведениям, располагающихся в ЗАО. 

    4) Средняя стоимость чашки капучино по округам Москвы:
        
- В среднем, цены за капучино в Москве варьируются от 151 р. до 191 р;
    
- Средняя стоимость капучино выше всего в таких округах , как ЗАО, ЦАО и ЮЗАО (190 р., 188 р. и 185 р. за чашку соответственно);
    
- В ЮАО и ЮВАО средняя цена за чашку капучино ниже всех - примерно 155 р., 152 р. соответственно.

 <b> РЕКОМЕНДАЦИИ : </b> 
    

    В целом, открытие кофейни является целесообразным и подходящим вариантом, ведь это третья по популярности заведений общепита в Москве категория: доля кофеен составляет 17% от всего количества заведений разных типов. Для сравнения: лидеры рынка по популярности, такие как кафе и рестораны, составляют по 29% и 24% соответственно.

    Кофейня - это в основном сетевой тип заведения, поэтому есть превосходная возможность создать целую сеть кофеен и распространиться по всему городу и даже за его пределами в будущем.
     
    Самый главный конкурент среди кофеен – сеть 'Шоколадница'. Эта кофейня вошла в топ-15 популярных сетей в Москве. 

    В центре расположено заведений в почти 2.5 раза больше, чем во втором по количеству заведений районе в Москве - СВАО, где локализовано 855 заведений.  

    Что касается самих кофеен, то больше всего кофеен расположено в ЦАО (407 штук) - это 1/3 от всех кофеен по всей Москве.  
    
    
 <b> Для открытия кофейни рекомендовано рассмотреть следующие округи: </b> 
    
    
    1. ЦАО: 
 
- Определенно стоит попробовать рассмотреть вариант открытия кофейни а ЦАО, ведь несмотря на высокий показатель конкуренции там, трафик людей в этой зоне крайне высокий. Более того, обычно людям удобнее встречаться в центре города, так что хорошие места общепита могут попасть в число рекомендаций;
    
- Более того, кофейню можно сделать круглосуточной, ведь именно в центре концетрация людей, желающих засидеться допоздна, очень большая. А конкурентов среди круглосуточных кофеен не много. 
    
    2. ЗАО: 
    
- Помимо ЦАО, много кофеен уже локализовано в ЗАО (140 штук). Там стоит попробовать открыть кофейню по той причине, что в результате проведенного анализа было выявлено, что в этом округе самый низкий средний рейтинг заведений сейчас. Это предоставляет большой шанс стать лучшей кофейней и выделиться из толпы;
   
- Средняя стоимость капучино в ЗАО самая высокая именно в ЗАО на данный момент (190 р.). Это сделало бы открытие новой кофейни там наиболее выгодным. 
    
    3. ЮЗАО: 
 
- Так же рекомендовано открытие кофейни в этом округе по причине того, что средний чек за чашку кофе там тоже один из самых высоких (185 р.);
    
- в ЮЗАО пока что небольшое количество кофеен.
    

    Совершенно точно не стоит рассматривать СЗАО для открытия кофейни, невзирая на тот факт, что по площади этот район является одним из крупных в Москве. Там уже в ходе анализа было выявлено наименьшее количество кофеен (всего 60 штук, что почти 7 раз меньше, чем в ЦАО).
