```python
def calculate_true_false_proportions(df):
    # Фильтруем только булевые столбцы
    bool_columns = data.select_dtypes(include=['bool'])

    # Создаем новый DataFrame для хранения долей
    proportions = pd.DataFrame(columns=['True', 'False'])

    for col in bool_columns:
        # Считаем количество True и False
        true_count = data[col].sum()  # Сумма True (1)
        false_count = len(data[col]) - true_count  # Количество False
        total_count = len(data[col])  # Общее количество значений

        # Рассчитываем доли
        true_proportion = true_count / total_count
        false_proportion = false_count / total_count

        # Добавляем результаты в новый DataFrame
        proportions.loc[col] = [true_proportion, false_proportion]

    return proportions

result = calculate_true_false_proportions(data)
display(result)
```

## Шаг 4. Построим модель прогнозирования оттока пользователей

Построим модель бинарной классификации пользователей, где целевой признак — факт оттока пользователя в следующем месяце:

### Разобъем данные на обучающую и валидационную выборку функцией train_test_split():

```python
# Определим функцию, которая будет выводить метрики accuracy, precision и recall:
def print_all_metrics(y_true, y_pred, y_proba, title='Метрики классификации'):
    print(title)
    print('\tAccuracy: {:.2f}'.format(accuracy_score(y_true, y_pred)))
    print('\tPrecision: {:.2f}'.format(precision_score(y_true, y_pred)))
    print('\tRecall: {:.2f}'.format(recall_score(y_true, y_pred)))
```

```python
# Разделим данные на признаки (X) и целевую переменную (y)
X = data.drop(columns=['churn'])
y = data['churn']
         
# Разделяем модель на обучающую и валидационную выборки:
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# Cоздадим объект класса StandardScaler и применим его к обучающей выборке:
scaler = StandardScaler()
X_train_st = scaler.fit_transform(X_train) # Обучаем scaler и трансформируем X_train
display(X_train_st[:5])
```

```python
#применяем стандартизацию к матрице признаков для тестовой выборки:
X_test_st = scaler.transform(X_test)
```

### Обучим модель на train-выборке двумя способами:

**1. Логистической регрессией:**

```python
# Создание модели:
lr_model = LogisticRegression(random_state=0, solver='liblinear')

# Обучим модель:
lr_model.fit(X_train_st, y_train)

# Прогнозирование на валидационной выборке:
lr_predictions = lr_model.predict(X_test_st)
lr_probabilities = lr_model.predict_proba(X_test_st)[:, 1]
```

```python
# Оценка метрик:
print_all_metrics(
    y_test,
    lr_predictions,
    lr_probabilities,
    title='Метрики для модели логистической регрессии:',
)
```

**2. Случайным лесом:**

```python
# Создание модели:
rf_model =  RandomForestClassifier(random_state=0)

# Обучаем модель на стандартизированных данных:
rf_model.fit(X_train_st, y_train)

# Прогнозирование на валидационной выборке:
y_pred = rf_model.predict(X_test)

# воспользуемся уже обученной моделью, чтобы сделать прогнозы
rf_predictions = rf_model.predict(X_test_st)
rf_probabilities = rf_model.predict_proba(X_test_st)[:, 1]
```

```python
# Оценка метрик. Dыведем все метрики:
print_all_metrics(
    y_test, 
    rf_predictions,
    rf_probabilities,
    title='Метрики для модели случайного леса:'
)
```

### Сравним метрики accuracy, precision и recall для обеих моделей на валидационной выборке:

```python
print_all_metrics(
    y_test,
    lr_predictions,
    lr_probabilities,
    title='Метрики для модели логистической регрессии:',
)

print("***" * 15)

print_all_metrics(
    y_test, 
    rf_predictions,
    rf_probabilities,
    title='Метрики для модели случайного леса:'
)
```

Логистическая регрессия также имеет более высокую точность, что означает, что среди всех положительных предсказаний она делает меньше ложных срабатываний.

3. Recall (полнота): логистическая регрессия: 0.83, случайный лес: 0.81.

Логистическая регрессия опять же превосходит случайный лес по полноте, что означает, что она лучше находит все истинные положительные случаи..

### Обучим модель кластеризации на основании алгоритма K-Means и спрогнозируем кластеры клиентов:

```python
# Задаём число кластеров, равное 5, и фиксируем значение random_state для воспроизводимости результата:
km = KMeans(n_clusters=5, random_state=0) 

# Применяем алгоритм к данным и формируем вектор кластеров
labels = km.fit_predict(X_sc)

# Добавление меток кластеров в DataFrame:
data['cluster_km'] = labels
data.head()
```

Все это свидетельствует о том, что клиенты класетров 0 и 4, которые как раз являются самыми лояльными клиентами фитнес-клуба, обладают лучшими показателями по признакам, описанным выше. А также, эти средние показатели не противоречат наблюдениям, сделанным ранее.

По приближенности к фитнес-клубу: видно ясно, что клиенты из всех кластеров, кроме кластера 2, проживают или работают близко к залу. Это подчеркивает важность дистанционной доступности к спортивному центру для посетителей.

Касательно клубов-партнеров: в кластеры 3 и 4 попали клиенты, которые не являются партнерами клуба, а вот в кластер 0 в основном попали клиенты, которые являются партнерами фитнес-клуба.

Что касается кластеров, то на кластер 0 приходится большее количество человек, которые пришли в клуб по акции «приведи друга», а в кластере 3 находится большинство клиентов, которые пришли не по акции.

Касательно групповых занятий: по графикам видно, что общее количество посетителей чуть больше среди тех, кто ходят не на групповые занятия. Но в целом посетители по всех кластерах ходят как на групповые, так и не на групповые тренировки в клубе.

Кластеры 0 и 4 характеризуются наиболее лояльными к клубу клиентами, когда как по остальным кластерам распределены одинаковое количество как лояльных, так и не лояльных посетителей клуба.

Касательно выручки от других услуг фитнес-центра: примерно одинаковый уровень трат представлен среди клиентов по всем кластерам, но достаточно много клиентов из кластера 3 тратят больше денег на доп. услуги клуба (судя по выбросам).

Клиенты, относящиеся к кластерам 4 и 0, являются наиболее “долгоживущими”, что говорит об их высокой лояльности и удовлетворенности предоставляемыми услугами.

Клиенты, относящиеся к 4 кластеру, отличаются высокой частотой посещений фитнес-клуба, а также стабильностью посещений на протяжении предыдущего месяца.

<b>Рекомендации для разработки стратегий по снижению оттока и повышению лояльности.:</b>

1) Так как средний возраст посетителей примерно 30 лет, стоит делать рекламу фитнес-клуба с акцентом на ценности данного возраста: забота о здоровье, высокая эффективность и уровень энергии, поддержание красивой подтянутой фигуры и т.д.;
    
2) сделать акцент на мотивации заниматься спортом более 2-3 раз в неделю;
    
3) разнообразить групповые тренировки, сделать их интереснее, чтобы клиенты чаще приходили заниматься на тернировки не только самостоятельно;
    
4) учитывать территориальный фактор:на отток клиентов влияет приближенность фитнес-центра к дому или месту работы;
    
5) улучшить и разнообразить число предложений дополнительных платных услуг в фитнес-центре;
    
6) мотивировать отдел продаж предлагать клиентам более долгосрочные абонементы;
    
7) так как клиенты из компаний-партнеров менее склонны к оттоку, то стоит лишь наращивать партнерскую программу. Хоть и акция "приведи друга" показала себя достаточно нейтрально и почти не влияла на лояльность клиентов, все стоит продолжать проводить эту акцию.
