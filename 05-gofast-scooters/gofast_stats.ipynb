Определим количество выбросов и их долю от всего датафрейма:

```python
# Вычисляем первый (Q1) и третий (Q3) квартиль и межквартильный размах (IQR):
Q1 = users_go['age'].quantile(0.25)
Q3 = users_go['age'].quantile(0.75)
IQR = Q3 - Q1

# Определяем границы для выбросов:
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Находим выбросы:
outliers = users_go[(users_go['age'] < lower_bound) | (users_go['age'] > upper_bound)]

# Считаем количество выбросов:
num_outliers = outliers.shape[0]

# Общее количество наблюдений :
total_count = users_go.shape[0]

# Считаем долю выбросов :
outlier_fraction = num_outliers / total_count


print(f"Количество выбросов: {num_outliers}")
print(f"Доля выбросов: {outlier_fraction:.2%}")
```

Аномальных значений в столбце 'age' всего 9 штук. Скорее всего, это редкие уникальные значения, не случайные и ошибочные, а просто редкие значения, ведь на самокате дейсвтиетльно могли прокатиться как пользователи 12 лет, так и  43 лет.

### Частота встречаемости городов

```python
#Создадим сводную табличку:

pivot_table_cities = users_go.pivot_table(
    index='city', values='user_id', aggfunc='count'
    ).sort_values(by='user_id', ascending=True)
```

```python
# Переименуем столбец user_id в number_of_users:
pivot_table_cities.rename(columns={'user_id': 'number_of_users'}, inplace=True)

# Вычислим общее количество пользователей:
total_users = pivot_table_cities['number_of_users'].sum()

# Добавим новый столбец с долями:
pivot_table_cities['share'] = round((pivot_table_cities['number_of_users'] / total_users) * 100, 1)

# Форматируем вывод долей в процентах без лишних нулей:
pivot_table_cities['share'] = pivot_table_cities['share'].apply(lambda x: f"{x:.1f}%")

# Отсортируем значения по убыванию по столбцу number_of_users:
pivot_table_cities.sort_values(by='number_of_users', ascending=False, inplace=True)



# Настраиваем стиль таблички:
styled_table = pivot_table_cities.style.set_table_styles([
    {'selector': 'th',  # Стилизация заголовков таблицы
     'props': [('background-color', '#A8E4A0'),  # Зеленый цвет
               ('color', 'black'),  # Цвет текста
               ('font-style', 'italic'),  # Курсивный текст
               ('font-weight', 'normal')]}  # Не жирный текст
])
display(styled_table)
```

```python
# Вычисляем первый (Q1) и третий (Q3) квартиль и межквартильный размах (IQR):
Q1 = rides_go['distance'].quantile(0.25)
Q3 = rides_go['distance'].quantile(0.75)
IQR = Q3 - Q1

# Определяем границы для выбросов:
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Находим выбросы:
outliers = rides_go[(rides_go['distance'] < lower_bound) | (rides_go['distance'] > upper_bound)]

# Считаем количество выбросов:
num_outliers = outliers.shape[0]

# Общее количество наблюдений :
total_count = rides_go.shape[0]

# Считаем долю выбросов :
outlier_fraction = num_outliers / total_count


print(f"Количество выбросов: {num_outliers}")
print(f"Доля выбросов: {outlier_fraction:.2%}")
```

## Шаг 6. Проверка гипотез

**Продакт-менеджеры сервиса хотят увеличить количество пользователей с подпиской. Для этого они будут проводить различные акции, но сначала нужно выяснить несколько важных моментов.**

### • Важно понять, тратят ли пользователи с подпиской больше времени на поездки? Если да, то пользователи с подпиской могут быть «выгоднее» для компании. 
Проверим гипотезу. Будем использовать исходные данные о продолжительности каждой сессии — отдельно для подписчиков и тех, у кого нет подписки.

***Решение*:** 

Для этого нам нужно посчитать среднее значение для продолжительности поездок у пользователй с подпиской и без.
Составим гипотезы, нулевую и альтурнативную, и проверим их:

• H₀ (нулевая гипотеза): μ₁ = μ₂ (среднее время поездки с подпиской равно среднему времени поездки без подписки)

• H₁ (альтернативная гипотеза, односторонняя): μ₁ > μ₂ (среднее время поездки с подпиской больше среднего времени поездки без подписки)

Нам необходимо использовать гипотезу о равенстве средних двух генеральных совокупностей:

```python
print(ultra_data.columns)
```

```python
# Извлекаем столбцы с длительностью:
ultra_duration = ultra_data['duration']
free_duration = free_data['duration']

# задаём уровень значимости:
alpha = 0.05

# Проводим t-тест:
results = st.ttest_ind(ultra_duration, free_duration, alternative='greater')

print(f'p-value: {results.pvalue}')
if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу.')
```

p-value в процентах = p-value × 100

1.2804098316645618e-34 × 100 = 1.2804098316645618 × 10⁻³² %

Это означает, что p-value составляет примерно 1.2804 × 10⁻³²%, что практически равно нулю.


```python
# Сверимся и посчитаем средние значения для двух столбцов:
average_durations = {
    'Dataset': ['ultra_data', 'free_data'],
    'Average Duration': [ultra_data['duration'].mean(), free_data['duration'].mean()]
}

# Создаем DataFrame:
average_durations_data = pd.DataFrame(average_durations)
average_durations_data['Average Duration'] = average_durations_data['Average Duration'].round(2)

display(average_durations_data)
```

Полученная вероятность p-value того, что среднее значение продолжительности поездки среди пользователей с подпиской равно среднему значению пользователей без нее крайне мало (1.2804098316645618e-34 - что практически равно нулю) и значительно меньше нашего заданного уровня значимости (0.05).

## Шаг 7. Распределения

### Отделу маркетинга GoFast поставили задачу: нужно провести акцию с раздачей промокодов на один бесплатный месяц подписки, в рамках которой как минимум 100 существующих клиентов должны продлить эту подписку. То есть по завершении периода действия подписки пользователь может либо отказаться от неё, либо продлить, совершив соответствующий платёж. 

По условию задачи сказано: " в рамках которой как минимум 100 существующих клиентов должны продлить эту подписку". Получается, акция проводилась ранее уже и по ее итогу более 10% пользователей продлевали подписку.

Нужно выяснить, какое минимальное количество промокодов нужно разослать, чтобы вероятность не выполнить план была примерно 5 %. 

Нужно подобрать параметры распределения, описывающего эту ситуацию, построить график распределения и сформулировать ответ на вопрос о количестве промокодов.
	
Нужно использовать биномиальное распределение, которое описывает указанную ситуацию. Затем подобрать подходящее значение параметра N для заданного параметра p=0.1 с помощью графиков и метода cdf().

**Дано:**

1)  p = 0.1  - вероятность успеха / продления подписки после получения промокода

2)  X — число клиентов, которые продлили подписку

3) N — количество разосланных промокодов (то, что мы ищем).

Нужно, чтобы вероятность того, что менее 100 клиентов продлят подписку, не превышала 5% (то есть P(X < 100) ≤ 0.05).


**Решение:**

1.  Формализация через биномиальное распределение:

Число клиентов, продливших подписку (X), подчиняется биномиальному распределению:

X ∼ Bin(N , p)

где p = 0.1 и N — количество разосланных промокодов.


Мы ищем минимальное N, при котором выполняется:

P(X<100)≤0.05

Это можно выразить через функцию распределения (CDF):

F(99)=P(X≤99)≤0.05

Нам нужно подобрать такое N, чтобы binom.cdf(99, N, 0.1) ≤ 0.05.

2. Подбор N с помощью Python:

Будем постепенно увеличивать N, пока вероятность P(X < 100) не станет ≤ 0.05.

```python
# Заданные параметры:

p = 0.1  # вероятность успеха - продления подписки после получения промокода (10% должны взять подписку)
threshold = 0.05  # допустимая вероятность провала
target_clients = 100  # целевое количество продлений
# или : k = 99    # Пороговое значение: 100 - 1 = 99

# Поиск минимального N, удовлетворяющего условию
N = 1000  # Начальное предположение/ количество попыток было 1000 до этого (по уловию, 100 человек составляли 10%)

# Запустим цикл:
for N in range(1000, 10000):  # Ограничиваем N сверху (например, 10000)
    if binom.cdf(target_clients - 1, N, p) <= threshold:
        break  # Как только нашли минимальное N, выходим из цикла
        
print(f"Минимальное количество промокодов, которое нужно разослать: {N}")
```

```python
# Построение графика распределения:

from scipy.stats import norm


N = 1161  # Найденное минимальное количество промокодов
p = 0.1  # Вероятность успешного продления подписки
target_clients = 100  # Минимальное количество успешных продлений

# Параметры нормального приближения:
mu = N * p  # Среднее значение
sigma = np.sqrt(N * p * (1 - p))  # Стандартное отклонение

# Создание диапазона значений X:
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Вычисление плотности вероятности нормального распределения:
y = norm.pdf(x, mu, sigma)

# Построение графика
plt.figure(figsize=(10, 6))
plt.plot(x, y, label="Аппроксимация биномиального через нормальное распределение", color="green")

# Вертикальная пунктирная линия для среднего (μ):
plt.axvline(mu, color="red", linestyle="--", label=f"Среднее (μ) = {mu:.1f}")

# Вертикальная линия для границы X = 99:
plt.axvline(target_clients - 1, color="blue", linestyle="--", label=f"Граница: X = 99")

# Добавление подписей и легенды:
plt.title("Нормальное распределение, аппроксимирующее биномиальное")
plt.xlabel("Количество успешных продлений (X)")
plt.ylabel("Плотность вероятности")
plt.legend()
plt.grid(True)

plt.show()
```

### Отдел маркетинга рассылает клиентам push-уведомления в мобильном приложении. Клиенты могут открыть его или не открывать. Известно, что уведомления открывают около 40 % получивших клиентов. Отдел планирует разослать 1 млн уведомлений. С помощью аппроксимации построим примерный график распределения и оценим вероятность того, что уведомление откроют не более 399,5 тыс. пользователей

Эта ситуация тоже описывается биномиальным распределением. Но считать каждое отдельное значение достаточно долго. Можно воспользоваться нормальной аппроксимацией биномиального распределения и cdf() для быстрой оценки.

```python
# Заданные параметры задачи:

N = 1000000  # Количество уведомлений
p = 0.4  # Вероятность открытия уведомления
target_clients = 399500  # Целевое количество пользователей

# Параметры нормального приближения:
mu = N * p  # Среднее значение
sigma = np.sqrt(N * p * (1 - p))  # Стандартное отклонение

# Создание диапазона значений X для графика:
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 10000)

# Вычисление плотности вероятности нормального распределения:
y = norm.pdf(x, mu, sigma)



# Построение графика:
plt.figure(figsize=(10, 6))
plt.plot(x, y, label="Аппроксимация биномиального через нормальное распределение", color="green")

# Вертикальная пунктирная линия для среднего (μ):
plt.axvline(mu, color="red", linestyle="--", label=f"Среднее (μ) = {mu:.1f}")

# Вертикальная линия для границы target_clients:
plt.axvline(target_clients, color="blue", linestyle="--", label=f"Граница: X = {target_clients}")

# Добавление подписей и легенды:
plt.title("Нормальное распределение, аппроксимирующее биномиальное")
plt.xlabel("Количество открытых уведомлений (X)")
plt.ylabel("Плотность вероятности")
plt.legend()
plt.grid(True)

plt.show()


# Оценка вероятности того, что уведомление откроют не более 399,5 тыс. пользователей

# Используем CDF (квантиль функции нормального распределения):
probability = norm.cdf(target_clients, mu, sigma)
print(f"Вероятность того, что уведомление откроют не более 399,5 тыс. пользователей: {probability:.4f}")
```
